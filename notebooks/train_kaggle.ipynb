```python
%%writefile data/prepare.py
"""RAGTruth -> sentence-level pairs.

Reads data/raw/{response,source_info}.jsonl (from github.com/ParticleMedia/RAGTruth) and writes data/processed/:
  sources.jsonl              one row per source: source_id, task_type, text (stored once, not per sentence)
  {train,val,test}.jsonl     one row per response sentence: source_id, sentence, label (1 = hallucinated), ...
  stats.json                 class balance and source lengths per split

Validation is carved out of train by source_id, because every source has responses from 6 LLMs and a random
split would leak. Standard library only, so it runs anywhere (laptop or Kaggle) in a few seconds.
"""

import argparse
import hashlib
import json
import re
import statistics
from collections import Counter
from pathlib import Path

ROOT = Path(__file__).resolve().parent
# A sentence ends at . ! or ? (optionally followed by a closing quote or bracket) and whitespace, or at a newline.
SENTENCE_END = re.compile(r"(?<=[.!?])[\"')\]]?\s+|\n+")
# Initials, common abbreviations and list numbers shouldn't end a sentence ("Dr. Smith", "J. K. Rowling", "1. Open").
NO_SPLIT_BEFORE = re.compile(r"(\b[A-Z]|\b(?:Mr|Mrs|Ms|Dr|St|Jr|Sr|vs|etc|e\.g|i\.e|No|Inc|Ltd|Co|U\.S)|^\s*\d{1,2})\.$")


def split_sentences(text: str) -> list[tuple[int, int]]:
    """Return (start, end) character offsets of each sentence, so annotated spans can be mapped onto them."""
    spans, start = [], 0
    for m in SENTENCE_END.finditer(text):
        if m.group().strip() == "" and NO_SPLIT_BEFORE.search(text[start : m.start()]):
            continue
        spans.append((start, m.start() + len(m.group().rstrip())))
        start = m.end()
    spans.append((start, len(text)))
    # Trim surrounding whitespace and drop empty pieces.
    out = []
    for s, e in spans:
        piece = text[s:e]
        if piece.strip():
            lead = len(piece) - len(piece.lstrip())
            out.append((s + lead, s + len(piece.rstrip())))
    return out


def source_text(src: dict) -> str:
    info = src["source_info"]
    if src["task_type"] == "QA":
        return f"Question: {info['question']}\n\n{info['passages']}"
    if src["task_type"] == "Data2txt":
        return json.dumps(info, ensure_ascii=False)
    return info  # Summary: the article itself


def split_of(row: dict, val_frac: float) -> str:
    if row["split"] == "test":
        return "test"
    bucket = int(hashlib.sha256(row["source_id"].encode()).hexdigest(), 16) % 1000
    return "val" if bucket < val_frac * 1000 else "train"


def main() -> None:
    ap = argparse.ArgumentParser()
    ap.add_argument("--raw", type=Path, default=ROOT / "raw")
    ap.add_argument("--out", type=Path, default=ROOT / "processed")
    ap.add_argument("--val-frac", type=float, default=0.1)
    args = ap.parse_args()
    args.out.mkdir(parents=True, exist_ok=True)

    sources = {}
    with open(args.raw / "source_info.jsonl") as f:
        for line in f:
            s = json.loads(line)
            sources[s["source_id"]] = {"source_id": s["source_id"], "task_type": s["task_type"], "text": source_text(s)}
    with open(args.out / "sources.jsonl", "w") as f:
        for s in sources.values():
            f.write(json.dumps(s, ensure_ascii=False) + "\n")

    rows = {"train": [], "val": [], "test": []}
    skipped = Counter()
    with open(args.raw / "response.jsonl") as f:
        for line in f:
            r = json.loads(line)
            if r["quality"] != "good":  # incorrect refusals and truncated responses
                skipped[r["quality"]] += 1
                continue
            split = split_of(r, args.val_frac)
            for i, (s, e) in enumerate(split_sentences(r["response"])):
                overlapping = [lab for lab in r["labels"] if lab["start"] < e and lab["end"] > s]
                rows[split].append({
                    "id": f"{r['id']}-{i}",
                    "response_id": r["id"],
                    "source_id": r["source_id"],
                    "task_type": sources[r["source_id"]]["task_type"],
                    "model": r["model"],
                    "sentence_idx": i,
                    "sentence": r["response"][s:e],
                    "label": int(bool(overlapping)),
                    "label_types": sorted({lab["label_type"] for lab in overlapping}),
                })

    ids = {k: {row["source_id"] for row in v} for k, v in rows.items()}
    assert not ids["train"] & ids["val"] and not ids["train"] & ids["test"] and not ids["val"] & ids["test"], \
        "source_id leaks across splits"

    stats = {"skipped_responses": dict(skipped)}
    for split, items in rows.items():
        with open(args.out / f"{split}.jsonl", "w") as f:
            for row in items:
                f.write(json.dumps(row, ensure_ascii=False) + "\n")
        src_words = [len(sources[sid]["text"].split()) for sid in ids[split]]
        pos = sum(row["label"] for row in items)
        stats[split] = {
            "sentences": len(items),
            "responses": len({row["response_id"] for row in items}),
            "sources": len(ids[split]),
            "hallucinated": pos,
            "hallucinated_frac": round(pos / len(items), 4),
            "hallucinated_frac_by_task": {
                t: round(sum(x["label"] for x in items if x["task_type"] == t)
                         / max(1, sum(x["task_type"] == t for x in items)), 4)
                for t in ("QA", "Summary", "Data2txt")
            },
            "source_words": {"median": statistics.median(src_words), "p95": sorted(src_words)[int(0.95 * len(src_words))],
                             "max": max(src_words)},
            "sentence_words_median": statistics.median(len(x["sentence"].split()) for x in items),
        }
    (args.out / "stats.json").write_text(json.dumps(stats, indent=2) + "\n")
    print(json.dumps(stats, indent=2))
    print("No source_id appears in more than one split.")


if __name__ == "__main__":
    main()
```

```python
%%writefile src/hd/data.py
"""Load the processed RAGTruth files written by data/prepare.py."""

import json
from pathlib import Path

DATA = Path(__file__).resolve().parents[2] / "data" / "processed"


def read_jsonl(path: Path) -> list[dict]:
    with open(path) as f:
        return [json.loads(line) for line in f]


def load_sources(data_dir: Path = DATA) -> dict[str, str]:
    return {s["source_id"]: s["text"] for s in read_jsonl(Path(data_dir) / "sources.jsonl")}


def load_split(split: str, data_dir: Path = DATA, limit: int | None = None) -> list[dict]:
    rows = read_jsonl(Path(data_dir) / f"{split}.jsonl")
    return rows[:limit] if limit else rows


def chunk_words(text: str, size: int = 250, overlap: int | None = None) -> list[str]:
    """Split a long source into overlapping word windows that fit a 512-token encoder alongside the sentence."""
    overlap = size // 5 if overlap is None else overlap
    if not 0 <= overlap < size:
        raise ValueError("overlap must be in [0, size)")
    words = text.split()
    if len(words) <= size:
        return [" ".join(words)]
    step = size - overlap
    return [" ".join(words[i : i + size]) for i in range(0, len(words) - overlap, step)]
```

```python
%%writefile src/hd/metrics.py
"""Metrics for the hallucinated class (label 1). Plain Python, so they run and test anywhere."""


def prf(labels: list[int], preds: list[int]) -> dict:
    tp = sum(1 for y, p in zip(labels, preds) if y and p)
    fp = sum(1 for y, p in zip(labels, preds) if not y and p)
    fn = sum(1 for y, p in zip(labels, preds) if y and not p)
    precision = tp / (tp + fp) if tp + fp else 0.0
    recall = tp / (tp + fn) if tp + fn else 0.0
    f1 = 2 * precision * recall / (precision + recall) if precision + recall else 0.0
    return {"precision": precision, "recall": recall, "f1": f1}


def best_threshold(labels: list[int], scores: list[float]) -> float:
    """Threshold t maximizing F1 when predicting hallucinated for score >= t."""
    pairs = sorted(zip(scores, labels), reverse=True)
    positives = sum(labels)
    tp = fp = 0
    best_f1, best_t = -1.0, 1.0
    for i, (s, y) in enumerate(pairs):
        tp += y
        fp += 1 - y
        if i + 1 < len(pairs) and pairs[i + 1][0] == s:
            continue  # only cut between distinct scores
        f1 = 2 * tp / (tp + fp + positives) if positives else 0.0
        if f1 > best_f1:
            best_f1, best_t = f1, s
    return best_t


def auroc(labels: list[int], scores: list[float]) -> float:
    """Area under the ROC curve via the Mann-Whitney rank statistic, with tied scores sharing an average rank."""
    order = sorted(range(len(scores)), key=lambda i: scores[i])
    ranks = [0.0] * len(scores)
    i = 0
    while i < len(order):
        j = i
        while j + 1 < len(order) and scores[order[j + 1]] == scores[order[i]]:
            j += 1
        for k in range(i, j + 1):
            ranks[order[k]] = (i + j) / 2 + 1
        i = j + 1
    n_pos = sum(labels)
    n_neg = len(labels) - n_pos
    if not n_pos or not n_neg:
        return float("nan")
    rank_sum = sum(r for r, y in zip(ranks, labels) if y)
    return (rank_sum - n_pos * (n_pos + 1) / 2) / (n_pos * n_neg)


def report(labels: list[int], scores: list[float], threshold: float) -> dict:
    out = prf(labels, [int(s >= threshold) for s in scores])
    out["auroc"] = auroc(labels, scores)
    out = {k: round(v, 4) for k, v in out.items()}
    out.update(n=len(labels), hallucinated=sum(labels))
    return out
```

```python
%%writefile src/hd/baselines.py
"""Zero-shot NLI baseline: a sentence is hallucinated when no chunk of its source entails it.

score = 1 - max over source chunks of P(entailment | chunk, sentence), so higher means more likely hallucinated.
No training. Run on a GPU (Kaggle): PYTHONPATH=src python -m hd.baselines --splits val test
"""

import argparse
import json
import time
from pathlib import Path

from hd.data import chunk_words, load_sources, load_split

NLI_MODEL = "cross-encoder/nli-deberta-v3-base"


class NLIScorer:
    def __init__(self, model_name: str = NLI_MODEL, batch_size: int = 64):
        import torch
        from transformers import AutoModelForSequenceClassification, AutoTokenizer

        self.torch = torch
        self.device = "cuda" if torch.cuda.is_available() else "cpu"
        self.tok = AutoTokenizer.from_pretrained(model_name)
        self.model = AutoModelForSequenceClassification.from_pretrained(model_name).to(self.device).eval()
        if self.device == "cuda":
            self.model.half()
        labels = {name.lower(): i for i, name in self.model.config.id2label.items()}
        self.entail_idx = labels["entailment"]
        self.batch_size = batch_size

    def __call__(self, pairs: list[tuple[str, str]]) -> list[float]:
        """P(entailment) for each (premise, hypothesis) pair. Pairs are batched by length to cut padding."""
        order = sorted(range(len(pairs)), key=lambda i: len(pairs[i][0]))
        probs = [0.0] * len(pairs)
        with self.torch.inference_mode():
            for b in range(0, len(order), self.batch_size):
                idx = order[b : b + self.batch_size]
                enc = self.tok([pairs[i][0] for i in idx], [pairs[i][1] for i in idx], truncation="only_first",
                               max_length=512, padding=True, return_tensors="pt").to(self.device)
                p = self.model(**enc).logits.float().softmax(-1)[:, self.entail_idx].tolist()
                for i, v in zip(idx, p):
                    probs[i] = v
        return probs


def score_rows(rows: list[dict], sources: dict[str, str], entailment, chunk_size: int = 250) -> tuple[list[float], int]:
    """Return (hallucination score per row, number of NLI pairs run)."""
    chunks = {sid: chunk_words(sources[sid], chunk_size) for sid in {r["source_id"] for r in rows}}
    pairs, owner = [], []
    for i, r in enumerate(rows):
        for c in chunks[r["source_id"]]:
            pairs.append((c, r["sentence"]))
            owner.append(i)
    best = [0.0] * len(rows)
    for i, p in zip(owner, entailment(pairs)):
        best[i] = max(best[i], p)
    return [1.0 - b for b in best], len(pairs)


def main() -> None:
    ap = argparse.ArgumentParser()
    ap.add_argument("--splits", nargs="+", default=["val", "test"])
    ap.add_argument("--limit", type=int, help="score only the first N sentences per split (smoke run)")
    ap.add_argument("--out", type=Path, default=Path("results"))
    args = ap.parse_args()
    args.out.mkdir(parents=True, exist_ok=True)

    sources = load_sources()
    scorer = NLIScorer()
    gpu = scorer.torch.cuda.get_device_name(0) if scorer.device == "cuda" else "cpu"
    timing = {"model": NLI_MODEL, "device": gpu}
    for split in args.splits:
        rows = load_split(split, limit=args.limit)
        start = time.perf_counter()
        scores, n_pairs = score_rows(rows, sources, scorer)
        seconds = time.perf_counter() - start
        with open(args.out / f"nli_{split}.jsonl", "w") as f:
            for r, s in zip(rows, scores):
                f.write(json.dumps({"id": r["id"], "task_type": r["task_type"], "model": r["model"],
                                    "label": r["label"], "score": round(s, 6)}) + "\n")
        timing[split] = {"sentences": len(rows), "nli_pairs": n_pairs, "seconds": round(seconds, 1),
                         "ms_per_sentence": round(1000 * seconds / len(rows), 2)}
        print(split, timing[split], flush=True)
    (args.out / "nli_timing.json").write_text(json.dumps(timing, indent=2) + "\n")


if __name__ == "__main__":
    main()
```

```python
%%writefile src/hd/model.py
"""ModernBERT pair classifier: [CLS] sentence [SEP] source [SEP] -> one logit, higher means more likely hallucinated.

The sentence goes first, so truncation only ever trims the tail of the source and never the claim being checked.
Packing and batching are plain Python (testable without torch); torch is imported only where tensors are built.
"""

import random

MODEL_NAME = "answerdotai/ModernBERT-base"


def pack(sent_ids: list[int], src_ids: list[int], max_len: int, cls_id: int, sep_id: int) -> list[int]:
    sent_ids = sent_ids[: max_len // 2]  # guard against a pathological sentence crowding out the whole source
    room = max_len - len(sent_ids) - 3
    return [cls_id, *sent_ids, sep_id, *src_ids[:room], sep_id]


class PairEncoder:
    """Token ids for (sentence, source) rows. Each source is tokenized once and shared by all its sentences."""

    def __init__(self, tokenizer, max_len: int = 2048):
        self.tok = tokenizer
        self.max_len = max_len

    def _ids(self, texts: list[str]) -> list[list[int]]:
        return self.tok(texts, add_special_tokens=False)["input_ids"]

    def encode(self, rows: list[dict], sources: dict[str, str]) -> tuple[list[list[int]], float]:
        """Return (packed ids per row, fraction of rows whose source was truncated)."""
        sids = sorted({r["source_id"] for r in rows})
        src = dict(zip(sids, self._ids([sources[s] for s in sids])))
        sents = self._ids([r["sentence"] for r in rows])
        seqs, cut = [], 0
        for r, s in zip(rows, sents):
            seq = pack(s, src[r["source_id"]], self.max_len, self.tok.cls_token_id, self.tok.sep_token_id)
            cut += len(seq) < min(len(s), self.max_len // 2) + len(src[r["source_id"]]) + 3
            seqs.append(seq)
        return seqs, cut / max(len(rows), 1)


def make_batches(lengths: list[int], token_budget: int, max_batch: int, shuffle: bool = False,
                 seed: int = 0) -> list[list[int]]:
    """Group example indices so each padded batch (size * longest) stays within token_budget.

    Similar lengths are grouped to cut padding. With shuffle, examples are sorted inside random pools of ~100
    batches and the batch order is shuffled, so each epoch sees different neighbours without sorting the whole set.
    """
    idx = list(range(len(lengths)))
    rng = random.Random(seed)
    if shuffle:
        rng.shuffle(idx)
        pool = 100 * max_batch
        idx = [i for p in range(0, len(idx), pool) for i in sorted(idx[p : p + pool], key=lambda i: lengths[i])]
    else:
        idx.sort(key=lambda i: lengths[i])
    batches, cur, longest = [], [], 0
    for i in idx:
        new_longest = max(longest, lengths[i])
        if cur and (len(cur) + 1 > max_batch or (len(cur) + 1) * new_longest > token_budget):
            batches.append(cur)
            cur, new_longest = [], lengths[i]
        cur.append(i)
        longest = new_longest
    if cur:
        batches.append(cur)
    if shuffle:
        rng.shuffle(batches)
    return batches


def collate(seqs: list[list[int]], pad_id: int):
    import torch

    width = max(len(s) for s in seqs)
    ids = torch.full((len(seqs), width), pad_id, dtype=torch.long)
    mask = torch.zeros((len(seqs), width), dtype=torch.long)
    for i, s in enumerate(seqs):
        ids[i, : len(s)] = torch.tensor(s)
        mask[i, : len(s)] = 1
    return ids, mask


def load_model(name: str = MODEL_NAME):
    from transformers import AutoConfig, AutoModelForSequenceClassification

    config = AutoConfig.from_pretrained(name, num_labels=1)
    if hasattr(config, "reference_compile"):  # ModernBERT would torch.compile and recompile for every batch shape
        config.reference_compile = False
    return AutoModelForSequenceClassification.from_pretrained(name, config=config, attn_implementation="sdpa")
```

```python
%%writefile src/hd/train.py
"""Fine-tune ModernBERT on RAGTruth sentence pairs with a plain PyTorch loop (no HF Trainer).

fp16 autocast + GradScaler (the T4 has no bf16), token-budget batches, gradient accumulation, AdamW with linear
warmup and decay, and a class-weighted BCE loss. The epoch with the best val F1 is kept and scored on test.
Writes results/ft_{val,test}.jsonl in the same format as the NLI baseline, so eval/evaluate.py --method ft works.

Kaggle smoke run (must overfit):  PYTHONPATH=src python -m hd.train --smoke 64 --epochs 15 --lr 1e-4 --accum 1 \
                                  --out /tmp/smoke --ckpt /tmp/smoke/ckpt
Full run:                         PYTHONPATH=src python -m hd.train
"""

import argparse
import json
import math
import time
from pathlib import Path

from hd.data import load_sources, load_split
from hd.metrics import best_threshold, report
from hd.model import MODEL_NAME, PairEncoder, collate, load_model, make_batches


def lr_factor(step: int, warmup: int, total: int) -> float:
    """Linear warmup to 1, then linear decay to 0."""
    if step < warmup:
        return (step + 1) / warmup
    return max(0.0, (total - step) / max(1, total - warmup))


def pos_weight(labels: list[int], mode: str) -> float:
    """BCE weight for the hallucinated class. 'sqrt' softens the full neg/pos ratio (~8 here), which tends to
    push scores toward 1 and hurt calibration; the decision threshold is tuned on val anyway."""
    pos = sum(labels)
    ratio = (len(labels) - pos) / max(pos, 1)
    return {"none": 1.0, "sqrt": math.sqrt(ratio), "ratio": ratio}[mode]


def smoke_sample(rows: list[dict], n: int) -> list[dict]:
    pos = [r for r in rows if r["label"]][: n // 2]
    neg = [r for r in rows if not r["label"]][: n - len(pos)]
    return pos + neg


class Trainer:
    def __init__(self, args):
        import torch
        from transformers import AutoTokenizer

        self.torch, self.args = torch, args
        torch.manual_seed(args.seed)
        self.device = "cuda" if torch.cuda.is_available() else "cpu"
        self.amp = self.device == "cuda"
        self.tok = AutoTokenizer.from_pretrained(args.model)
        self.enc = PairEncoder(self.tok, args.max_len)
        self.model = load_model(args.model).to(self.device)
        if args.grad_ckpt:
            self.model.gradient_checkpointing_enable()

    def batches(self, seqs, shuffle, seed=0):
        return make_batches([len(s) for s in seqs], self.args.token_budget, self.args.max_batch, shuffle, seed)

    def logits(self, seqs, idx):
        ids, mask = collate([seqs[i] for i in idx], self.tok.pad_token_id)
        with self.torch.autocast("cuda", dtype=self.torch.float16, enabled=self.amp):
            return self.model(input_ids=ids.to(self.device), attention_mask=mask.to(self.device)).logits.squeeze(-1)

    def predict(self, seqs) -> list[float]:
        self.model.eval()
        out = [0.0] * len(seqs)
        with self.torch.inference_mode():
            for idx in self.batches(seqs, shuffle=False):
                for i, p in zip(idx, self.logits(seqs, idx).float().sigmoid().tolist()):
                    out[i] = p
        return out

    def probe_memory(self):
        """One forward/backward on a budget-filling batch of max-length inputs, so an OOM shows up in seconds."""
        if not self.amp:
            return None
        n = max(1, min(self.args.max_batch, self.args.token_budget // self.args.max_len))
        seqs = [[self.tok.cls_token_id] * self.args.max_len] * n
        self.model.train()
        self.logits(seqs, list(range(n))).float().sum().backward()
        self.model.zero_grad(set_to_none=True)
        peak = self.torch.cuda.max_memory_allocated() / 2**30
        self.torch.cuda.reset_peak_memory_stats()
        return round(peak, 2)

    def fit(self, train_seqs, train_labels, val_seqs, val_labels, ckpt: Path) -> list[dict]:
        torch, a = self.torch, self.args
        n_batches = len(self.batches(train_seqs, shuffle=True))
        total = math.ceil(n_batches / a.accum) * a.epochs
        warmup = max(1, int(a.warmup * total))
        no_decay = [p for n, p in self.model.named_parameters() if p.ndim < 2 or "norm" in n.lower()]
        decay = [p for n, p in self.model.named_parameters() if not (p.ndim < 2 or "norm" in n.lower())]
        opt = torch.optim.AdamW([{"params": decay, "weight_decay": a.weight_decay},
                                 {"params": no_decay, "weight_decay": 0.0}], lr=a.lr)
        sched = torch.optim.lr_scheduler.LambdaLR(opt, lambda s: lr_factor(s, warmup, total))
        scaler = torch.amp.GradScaler("cuda", enabled=self.amp)
        loss_fn = torch.nn.BCEWithLogitsLoss(pos_weight=torch.tensor(pos_weight(train_labels, a.pos_weight),
                                                                     device=self.device))
        labels = torch.tensor(train_labels, dtype=torch.float)
        print(f"{len(train_seqs)} train examples, {n_batches} batches/epoch, {total} optimizer steps", flush=True)

        log, best_f1, step = [], -1.0, 0
        for epoch in range(a.epochs):
            self.model.train()
            start, tokens, run_loss, run_n = time.perf_counter(), 0, 0.0, 0
            batches = self.batches(train_seqs, shuffle=True, seed=a.seed + epoch)
            for b, idx in enumerate(batches):
                loss = loss_fn(self.logits(train_seqs, idx).float(), labels[idx].to(self.device))
                scaler.scale(loss / a.accum).backward()
                run_loss, run_n = run_loss + loss.item(), run_n + 1
                tokens += sum(len(train_seqs[i]) for i in idx)
                if (b + 1) % a.accum == 0 or b + 1 == len(batches):
                    scaler.unscale_(opt)
                    torch.nn.utils.clip_grad_norm_(self.model.parameters(), 1.0)
                    scaler.step(opt)
                    scaler.update()
                    opt.zero_grad(set_to_none=True)
                    sched.step()
                    step += 1
                    if step % a.log_every == 0:
                        secs = time.perf_counter() - start
                        eta = secs / (b + 1) * (len(batches) - b - 1) / 60
                        print(f"epoch {epoch} step {step}/{total} loss {run_loss / run_n:.4f} "
                              f"lr {sched.get_last_lr()[0]:.2e} {tokens / secs:.0f} tok/s, epoch eta {eta:.0f} min",
                              flush=True)
                        run_loss, run_n = 0.0, 0
            train_secs = time.perf_counter() - start
            scores = self.predict(val_seqs)
            t = best_threshold(val_labels, scores)
            entry = {"epoch": epoch, "train_minutes": round(train_secs / 60, 1),
                     "train_tok_per_s": round(tokens / train_secs), "threshold": round(t, 6),
                     "val": report(val_labels, scores, t)}
            if self.amp:
                entry["peak_gpu_gb"] = round(torch.cuda.max_memory_allocated() / 2**30, 2)
            print(json.dumps(entry), flush=True)
            log.append(entry)
            if entry["val"]["f1"] > best_f1:
                best_f1 = entry["val"]["f1"]
                self.model.save_pretrained(ckpt)
                self.tok.save_pretrained(ckpt)
        return log


def write_preds(path: Path, rows: list[dict], scores: list[float]) -> None:
    with open(path, "w") as f:
        for r, s in zip(rows, scores):
            f.write(json.dumps({"id": r["id"], "task_type": r["task_type"], "model": r["model"],
                                "label": r["label"], "score": round(s, 6)}) + "\n")


def main() -> None:
    ap = argparse.ArgumentParser()
    ap.add_argument("--model", default=MODEL_NAME)
    ap.add_argument("--max-len", type=int, default=2048)
    ap.add_argument("--token-budget", type=int, default=8192, help="max padded tokens per micro-batch")
    ap.add_argument("--max-batch", type=int, default=32)
    ap.add_argument("--accum", type=int, default=4)
    ap.add_argument("--epochs", type=int, default=2)
    ap.add_argument("--lr", type=float, default=3e-5)
    ap.add_argument("--warmup", type=float, default=0.06, help="fraction of optimizer steps")
    ap.add_argument("--weight-decay", type=float, default=0.01)
    ap.add_argument("--pos-weight", choices=["none", "sqrt", "ratio"], default="sqrt")
    ap.add_argument("--grad-ckpt", action="store_true")
    ap.add_argument("--limit", type=int, help="use only the first N train and val rows")
    ap.add_argument("--smoke", type=int, help="train and validate on the same N balanced train rows (must overfit)")
    ap.add_argument("--seed", type=int, default=0)
    ap.add_argument("--log-every", type=int, default=100)
    ap.add_argument("--out", type=Path, default=Path("results"))
    ap.add_argument("--ckpt", type=Path, default=Path("checkpoints/ft"))
    args = ap.parse_args()
    args.out.mkdir(parents=True, exist_ok=True)

    sources = load_sources()
    if args.smoke:
        train = val = smoke_sample(load_split("train"), args.smoke)
    else:
        train, val = load_split("train", limit=args.limit), load_split("val", limit=args.limit)
    tr = Trainer(args)
    train_seqs, train_cut = tr.enc.encode(train, sources)
    val_seqs, _ = tr.enc.encode(val, sources)
    info = {"config": {k: str(v) for k, v in vars(args).items()},
            "device": tr.torch.cuda.get_device_name(0) if tr.amp else "cpu",
            "train_truncated_frac": round(train_cut, 4), "probe_peak_gpu_gb": tr.probe_memory()}
    print(json.dumps(info), flush=True)

    info["epochs"] = tr.fit(train_seqs, [r["label"] for r in train], val_seqs, [r["label"] for r in val], args.ckpt)
    tr.model = load_model(args.ckpt).to(tr.device)  # best epoch by val F1
    write_preds(args.out / "ft_val.jsonl", val, tr.predict(val_seqs))
    if not args.smoke:
        test = load_split("test")
        test_seqs, _ = tr.enc.encode(test, sources)
        start = time.perf_counter()
        scores = tr.predict(test_seqs)
        info["test_ms_per_sentence"] = round(1000 * (time.perf_counter() - start) / len(test), 2)
        write_preds(args.out / "ft_test.jsonl", test, scores)
    (args.out / "ft_train_log.json").write_text(json.dumps(info, indent=2) + "\n")


if __name__ == "__main__":
    main()
```

```python
%%writefile eval/evaluate.py
"""Metrics from saved predictions, so every reported number can be reproduced from files in results/.

The threshold is picked on validation F1 and then applied unchanged to test.
Usage: python eval/evaluate.py --method nli   (reads results/nli_{val,test}.jsonl, writes results/nli_metrics.json)
"""

import argparse
import json
import sys
from pathlib import Path

sys.path.insert(0, str(Path(__file__).resolve().parents[1] / "src"))
from hd.data import read_jsonl  # noqa: E402
from hd.metrics import best_threshold, report  # noqa: E402


def grouped(rows: list[dict], key: str, threshold: float) -> dict:
    out = {}
    for value in sorted({r[key] for r in rows}):
        sub = [r for r in rows if r[key] == value]
        out[value] = report([r["label"] for r in sub], [r["score"] for r in sub], threshold)
    return out


def main() -> None:
    ap = argparse.ArgumentParser()
    ap.add_argument("--method", required=True)
    ap.add_argument("--results", type=Path, default=Path("results"))
    args = ap.parse_args()

    val = read_jsonl(args.results / f"{args.method}_val.jsonl")
    test = read_jsonl(args.results / f"{args.method}_test.jsonl")
    t = best_threshold([r["label"] for r in val], [r["score"] for r in val])
    metrics = {
        "method": args.method,
        "threshold_from_val": round(t, 6),
        "val": report([r["label"] for r in val], [r["score"] for r in val], t),
        "test": report([r["label"] for r in test], [r["score"] for r in test], t),
        "test_by_task": grouped(test, "task_type", t),
        "test_by_llm": grouped(test, "model", t),
    }
    (args.results / f"{args.method}_metrics.json").write_text(json.dumps(metrics, indent=2) + "\n")
    print(json.dumps(metrics, indent=2))


if __name__ == "__main__":
    main()
```
